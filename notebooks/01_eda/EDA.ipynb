```python
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np

# Load salah satu file NetCDF CMEMS
ds = xr.open_dataset("/workspace/data/raw/cmems_reanalysis/your_file.nc")

# Cek struktur data
print(ds)
print("\nVariabel:", list(ds.data_vars))
print("Koordinat:", list(ds.coords))
print("Dimensi:", dict(ds.dims))

# Plot SST
ds['thetao'].isel(time=0, depth=0).plot(
    cmap='RdYlBu_r',
    figsize=(10, 6)
)
plt.title('SST Laut Banda — snapshot pertama')
plt.savefig('/workspace/reports/figures/eda_sst_snapshot.png', dpi=150)
plt.show()
```
